```python
import os
import pandas as pd
from PIL import Image
import glob

# Define the directories containing the data
data_dir = '../images/'
csv_dir = '../csv/'

# Initialize lists to store data
csv_data = []
images = []

# Use glob to list all CSV files in the csv directory
csv_files = glob.glob(os.path.join(csv_dir, '*.csv'))

# Iterate over the CSV files and load each one individually
for csv_file in csv_files:
    df = pd.read_csv(csv_file)
    csv_data.append(df)
    print(f"Loaded CSV file: {csv_file}")
    print(df.head())

# Use glob to list all files in the data directory and subdirectories
files = glob.glob(os.path.join(data_dir, '**'), recursive=True)

# Iterate over the files
for file_path in files:
    # Check if the file is a JPEG image
    if file_path.endswith('.jpeg'):
        img = Image.open(file_path)
        images.append(img)

# Display the number of images loaded
print(f"Number of images loaded: {len(images)}")
```

```python
df1 = csv_data[1]
df2 = csv_data[2]
# df3 = df1.merge(df2, how='left', left_on=['Bon_ID', 'Artikelnummer'], right_on=['Bon_ID', 'Artikelnr'])
df3 = df1.copy()

# Convert Menge_Rückläufer and Prozent_Rückläufer to float
df2['Menge_Rückläufer'] = df2['Menge_Rückläufer'].apply(lambda x: float(str(x).replace(',', '.')))
df2['Prozent_Rückläufer'] = df2['Prozent_Rückläufer'].apply(lambda x: float(str(x).replace(',', '.')))

# Add the new columns to df3
df3['Menge_Rückläufer'] = 0
df3['Prozent_Rückläufer'] = 0
df3['Menge_Rückläufer'] = df3['Menge_Rückläufer'].astype(float)
df3['Prozent_Rückläufer'] = df3['Prozent_Rückläufer'].astype(float)
# First, add the values for non-NaN Artikelnr
for _, row in df2.iterrows():
    bon_id = int(row['Bon_ID'])
    artikelnr = row['Artikelnr']
    menge_ruecklaeufer = row['Menge_Rückläufer']
    prozent_ruecklaeufer = row['Prozent_Rückläufer']
    
    if pd.notna(artikelnr):
        # If Artikelnr is not NaN, add the values to the corresponding rows in df1
        df3.loc[(df3['Bon_ID'] == bon_id) & (df3['Artikelnummer'] == artikelnr), 'Menge_Rückläufer'] += menge_ruecklaeufer
        df3.loc[(df3['Bon_ID'] == bon_id) & (df3['Artikelnummer'] == artikelnr), 'Prozent_Rückläufer'] += prozent_ruecklaeufer

# Then, distribute the values for NaN Artikelnr
for _, row in df2.iterrows():
    bon_id = int(row['Bon_ID'])
    artikelnr = row['Artikelnr']
    menge_ruecklaeufer = row['Menge_Rückläufer']
    prozent_ruecklaeufer = row['Prozent_Rückläufer']
    
    if pd.isna(artikelnr):
        # If Artikelnr is NaN, distribute the values among the rest of the Artikelnr found in the Bon_ID
        df1_subset = df3[(df3['Bon_ID'] == bon_id) & (df3['Menge_Rückläufer'] == 0) & (df3['Prozent_Rückläufer'] == 0)]
        num_existing_artikelnr = len(df1_subset)
        if num_existing_artikelnr > 0:
            df3.loc[(df3['Bon_ID'] == bon_id) & (df3['Menge_Rückläufer'] == 0), 'Menge_Rückläufer'] += menge_ruecklaeufer / num_existing_artikelnr
            df3.loc[(df3['Bon_ID'] == bon_id) & (df3['Prozent_Rückläufer'] == 0), 'Prozent_Rückläufer'] += prozent_ruecklaeufer / num_existing_artikelnr

# Display the updated DataFrame
print(df3.head())
```
